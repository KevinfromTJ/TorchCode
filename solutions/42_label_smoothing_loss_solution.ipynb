# 🟡 Solution: Label Smoothing Loss

Reference solution for label-smoothed cross-entropy.

```python
import torch
import torch.nn.functional as F
```

```python
# ✅ SOLUTION

def label_smoothing_loss(logits: torch.Tensor, targets: torch.Tensor, smoothing: float = 0.1) -> torch.Tensor:
    num_classes = logits.shape[-1]
    log_probs = F.log_softmax(logits, dim=-1)
    target_probs = torch.full_like(log_probs, smoothing / num_classes)
    target_probs.scatter_(1, targets.unsqueeze(1), (1.0 - smoothing) + smoothing / num_classes)
    return -(target_probs * log_probs).sum(dim=-1).mean()
```

```python
# Verify
logits = torch.randn(4, 5)
targets = torch.tensor([0, 1, 2, 3])
print('Loss:', label_smoothing_loss(logits, targets, smoothing=0.1))
print('Ref: ', F.cross_entropy(logits, targets, label_smoothing=0.1))
```
