```python
import torch
```

```python
# ✅ SOLUTION

def gradient_descent_minimize(f, x0, lr=0.1, steps=100):
    x = x0.clone().detach().requires_grad_(True)
    for _ in range(steps):
        y = f(x)
        if x.grad is not None:
            x.grad.zero_()
        y.backward()
        with torch.no_grad():
            x -= lr * x.grad
    return x.detach()
```

```python
# Demo
center = torch.tensor([1.0, -2.0, 0.5])
f = lambda x: ((x - center) ** 2).sum()
x = gradient_descent_minimize(f, torch.zeros(3), lr=0.1, steps=300)
print('argmin:', x, '(expected', center.tolist(), ')')
```
