# Solution: GPU Batch Scheduling (No Discards)

Reference solution — sort + greedy sweep.

```python
from typing import List
```

```python
# ✅ SOLUTION

class Solution:
    def min_batches(self, lengths: List[int], k: int, c: int) -> int:
        a = sorted(lengths)
        n = len(a)
        batches = 0
        i = 0
        while i < n:                       # each pass opens exactly one batch
            batches += 1
            start = a[i]                   # smallest length in this batch
            count = 0
            # grow while the padding gap stays within K and capacity allows
            while i < n and a[i] - start <= k and count < c:
                count += 1
                i += 1
        return batches
```

```python
# Demo
sol = Solution()
print(sol.min_batches([10, 11, 15, 16, 17], 1, 2))  # 3
print(sol.min_batches([1, 100, 2, 200], 10, 3))     # 3
print(sol.min_batches([1, 2, 3, 4, 5, 6], 100, 3))  # 2
```
