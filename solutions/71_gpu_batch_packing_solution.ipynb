# Solution: GPU Batch Packing (Fixed Token Buffer)

Reference solution — single next-fit sweep in arrival order.

```python
from typing import List
```

```python
# ✅ SOLUTION

class Solution:
    def min_batches(self, lengths: List[int], b: int, g: int) -> int:
        batches = 0
        used = -1                       # tokens used in the open batch; -1 = none open
        for length in lengths:
            # open a new batch if none is active or the request no longer fits
            if used < 0 or used + g + length > b:
                batches += 1
                used = length           # first request in the batch pays no gap
            else:
                used += g + length      # +G separator, then the request itself
        return batches
```

```python
# Demo
sol = Solution()
print(sol.min_batches([3, 3, 3], 10, 1))   # 2
print(sol.min_batches([4, 4], 9, 1))       # 1  (4+1+4 == 9)
print(sol.min_batches([4, 1, 4, 1], 5, 0)) # 2  (order matters vs sorted -> 3)
```
