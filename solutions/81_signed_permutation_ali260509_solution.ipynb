# Solution: Signed Permutation (Sum Zero)

Reference solution.

## 解析

**结论：当且仅当 `n % 4 in {1, 2}` 时无解；否则用四元组 `(x, -(x+1), -(x+2), x+3)`（每组和为 0）拼出答案。**

### 可行性
给每个数配 `+/-` 号后，总和与 `S = 1+2+...+n = n(n+1)/2` 同奇偶（改一个符号使总和变化 `2*|a_i|`，偶数）。要凑出 0 必须 `S` 为偶数。`S` 为偶 ⇔ `n % 4 in {0, 3}`。所以 `n % 4 in {1, 2}` 无解。

### 构造
连续四个数 `x, x+1, x+2, x+3` 取符号 `+,-,-,+` 时和为 `x-(x+1)-(x+2)+(x+3)=0`。
- `n % 4 == 0`：从 `x=1` 起每 4 个一组，正好铺满。
- `n % 4 == 3`：先放 `[1, 2, -3]`（和为 0），剩下从 `x=4` 开始的个数是 4 的倍数，用四元组铺满。

### 复杂度
`O(n)` 构造，`O(n)` 空间。

```python
from typing import List
```

```python
# ✅ SOLUTION

class Solution:
    def construct(self, n: int) -> List[int]:
        if n % 4 in (1, 2):
            return []
        parts = []
        start = 1
        if n % 4 == 3:
            parts += [1, 2, -3]   # sums to 0
            start = 4
        for x in range(start, n + 1, 4):
            parts += [x, -(x + 1), -(x + 2), x + 3]   # sums to 0
        return parts
```

```python
# Demo
sol = Solution()
for n in [2, 3, 4, 7, 8]:
    r = sol.construct(n)
    print(n, '->', r, '(sum', sum(r), ')')
```
