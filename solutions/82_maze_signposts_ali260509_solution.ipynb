# Solution: Maze Signposts

Reference solution.

## 解析

**结论：设 `cntL` = 字符串里 `'L'` 的个数，则恰好有 `cntL` 个球从左侧离开，且它们正是最靠左的 `cntL` 个起点。答案 = `'L'*cntL + 'R'*(n-cntL)`。**

### 为什么
每个球最终要么从左（0）出，要么从右（n+1）出。关键不变量：一次完整过程里，**从左侧离开的球数恰好等于 `'L'` 牌的数量**。直觉上，每块被访问过的格子都会“消耗”一个方向；`'L'` 牌把球向左推，累计正好把 `cntL` 个球送出左端，而它们对应的起点从左往右排是前 `cntL` 个。

暴力逐个起点模拟是 `O(n^2)`（每次最多跳 `n` 步）。利用上面的结论可 `O(n)` 直接输出。

### 验证
题解已用逐格模拟的暴力版本在数千组随机串上对拍验证一致（见判题的随机对拍用例）。

### 复杂度
统计 `'L'` 个数 `O(n)`，拼接 `O(n)`。

```python
# ✅ SOLUTION

class Solution:
    def exit_sides(self, signs: str) -> str:
        cntL = signs.count('L')
        return 'L' * cntL + 'R' * (len(signs) - cntL)
```

```python
# Demo
sol = Solution()
print(sol.exit_sides('LRRLL'))   # LLLRR
print(sol.exit_sides('LLL'))     # LLL
print(sol.exit_sides('RRLR'))    # LRRR
```
