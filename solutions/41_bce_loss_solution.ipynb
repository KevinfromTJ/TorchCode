# 🟢 Solution: Binary Cross-Entropy Loss

Reference solution for numerically stable BCE-with-logits.

```python
import torch
```

```python
# ✅ SOLUTION

def bce_loss(logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
    loss = torch.clamp(logits, min=0) - logits * targets + torch.log1p(torch.exp(-logits.abs()))
    return loss.mean()
```

```python
# Verify
logits = torch.tensor([2.0, -1.0, 0.5, -3.0])
targets = torch.tensor([1.0, 0.0, 1.0, 0.0])
print('Loss:', bce_loss(logits, targets))
print('Ref: ', torch.nn.functional.binary_cross_entropy_with_logits(logits, targets))
```
