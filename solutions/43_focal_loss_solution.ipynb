# 🟡 Solution: Focal Loss

Reference solution for binary focal loss on logits.

```python
import torch
import torch.nn.functional as F
```

```python
# ✅ SOLUTION

def focal_loss(logits: torch.Tensor, targets: torch.Tensor, gamma: float = 2.0, alpha: float | None = None) -> torch.Tensor:
    bce = F.binary_cross_entropy_with_logits(logits, targets, reduction='none')
    probs = torch.sigmoid(logits)
    p_t = probs * targets + (1.0 - probs) * (1.0 - targets)
    loss = (1.0 - p_t) ** gamma * bce
    if alpha is not None:
        alpha_t = alpha * targets + (1.0 - alpha) * (1.0 - targets)
        loss = alpha_t * loss
    return loss.mean()
```

```python
# Verify
logits = torch.tensor([2.0, -2.0, 0.0, -1.5])
targets = torch.tensor([1.0, 0.0, 1.0, 0.0])
print('Loss:', focal_loss(logits, targets, gamma=2.0))
print('BCE :', F.binary_cross_entropy_with_logits(logits, targets))
```
