# Solution: Side-Scroller Coin Catching

Reference solution — 2D partial order + strict longest increasing subsequence.

```python
from typing import List
from bisect import bisect_left
```

```python
# ✅ SOLUTION

class Solution:
    def max_coins(self, coins: List[List[int]]) -> int:
        # T_B - T_A > |X_B - X_A|  <=>  (T-X) and (T+X) both strictly increase
        points = [(t - x, t + x) for t, x in coins]
        # sort by u asc; ties on u -> v desc so equal-u coins can't chain together
        points.sort(key=lambda p: (p[0], -p[1]))
        tails = []                       # strict LIS over v
        for _, v in points:
            pos = bisect_left(tails, v)  # first tail >= v -> keeps it strict
            if pos == len(tails):
                tails.append(v)
            else:
                tails[pos] = v
        return len(tails)
```

```python
# Demo
print(Solution().max_coins([[1, 2], [5, 5], [2, 3], [7, 4], [6, 8]]))  # 3
print(Solution().max_coins([[1, 0], [2, 0], [3, 0], [4, 0]]))          # 4
```
