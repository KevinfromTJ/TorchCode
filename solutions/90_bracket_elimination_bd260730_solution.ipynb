# 参考解法：括号分层消除

Reference solution.

## 解析

**结论：字符消失的轮次由它的嵌套深度唯一决定，一次扫描即可求出，无需真的做删除。**

### 深度即轮次
最内层的括号对总是最先被删除。一个字符所在的嵌套深度越深，就越早消失。设最大深度为 `D`，深度为 `d` 的字符会在第 `D - d + 1` 轮被删除。

### 如何取深度
扫描时遇 `'('` 先 `depth += 1` 再记录（左括号属于它开启的那一层），遇 `')'` 先记录 `depth` 再 `depth -= 1`（右括号与其配对左括号同层）。

### 验证
已用逐轮真实模拟删除的暴力实现在数千组随机合法括号串上对拍一致，并复现全部官方样例。

### 复杂度
一次线性扫描，`O(n)` 时间、`O(n)` 输出空间。

```python
# ✅ SOLUTION

from typing import List

class Solution:
    def rounds(self, s: str) -> List[int]:
        depth = 0
        rec = []
        for ch in s:
            if ch == '(':
                depth += 1
                rec.append(depth)   # '(' belongs to the layer it opens
            else:
                rec.append(depth)       # ')' matches its '(', same layer
                depth -= 1
        D = max(rec)
        return [D - d + 1 for d in rec]
```

```python
sol = Solution()
print(sol.rounds("(()())"))       # [2, 1, 1, 1, 1, 2]
print(sol.rounds("(()((())))"))   # [4, 3, 3, 3, 2, 1, 1, 2, 3, 4]
```
