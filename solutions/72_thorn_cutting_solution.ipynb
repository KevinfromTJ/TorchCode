# Solution: Thorn Forest Optimal Cutting

Reference solution — sort descending + greedy.


## 解析

**结论：把坚硬度从大到小排序，依次贪心尝试。**

失败的尝试没有任何收益、只会白白消耗 1 点锋利度，所以最优策略绝不会主动去砍一株砍不断的荆棘。于是问题变成：**选出尽量多的荆棘并安排顺序**，使得第 `j` 次成功砍断时锋利度满足 `K - (j-1) >= a_i`。

既然每成功一次锋利度就降 1，越硬的荆棘越应该趁锋利度高时先砍。因此：

- 将 `a` 从大到小排序；
- 维护已成功次数 `cnt`，当前锋利度即 `K - cnt`；
- 若 `K - cnt >= a_i` 就砍断它，`cnt += 1`；否则跳过（后面的更软，但锋利度只会更低，它今后也不可能被砍）。

**贪心正确性**：先处理更硬的荆棘，若它能砍则先砍不劣（软的以后更容易砍）；若它不能砍，随着成功次数增加锋利度只减不增，它以后更不可能砍，可直接跳过。

**复杂度**：排序 `O(n log n)`，扫描 `O(n)`；空间 `O(n)`。

```python
from typing import List
```

```python
# ✅ SOLUTION

class Solution:
    def max_cuts(self, a: List[int], k: int) -> int:
        a = sorted(a, reverse=True)   # hardest first
        cnt = 0
        for x in a:
            if k - cnt >= x:          # current sharpness is k - cnt
                cnt += 1
        return cnt
```

```python
# Demo
sol = Solution()
print(sol.max_cuts([2, 1, 4, 10, 3], 5))  # 4
print(sol.max_cuts([10, 10], 1))          # 0
print(sol.max_cuts([3, 3, 3], 3))         # 1
```
