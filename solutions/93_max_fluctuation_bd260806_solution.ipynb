# 参考解法：波动贡献最大化

Reference solution.

## 解析

**结论：最优划分就是不划分。答案为 `(max(v) - min(v)) * len(v)`。**

### 上界
任一段的极差 `(max - min)` 都不超过全局极差 `(M - m)`，因此该段贡献 `≤ (M - m) * 段长`。把所有段相加，段长之和为 `n`，于是任意划分的总贡献 `≤ (M - m) * n`。

### 上界可达
把整个序列作为唯一一段，其贡献恰为 `(M - m) * n`，达到上界。因此最大值就是它。

### 复杂度
一次遍历求最大最小，`O(n)`。数值可能较大，用 64 位整数（Python 原生大整数无溢出）。

### 验证
已用区间 DP 暴力在数千组随机序列上对拍一致，并复现官方样例。

```python
# ✅ SOLUTION

from typing import List

class Solution:
    def max_total(self, v: List[int]) -> int:
        return (max(v) - min(v)) * len(v)   # one whole segment is optimal
```

```python
sol = Solution()
print(sol.max_total([2, 5, 1]))      # 12
print(sol.max_total([1, 5, 2, 4]))   # 16
print(sol.max_total([3, 3, 3]))      # 0
```
