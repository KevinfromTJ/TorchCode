# Solution: Diagonal Traversal (Bottom-Left)

Reference solution.

## 解析

**结论：矩阵共有 `m+n-1` 条反对角线，从左下角那条开始逐条输出，奇偶条交替扫描方向。同一条对角线上 `i - j` 为定值 `d`。**

### 对角线编号
以 `g = 0` 表示从左下角出发的第一条对角线。同一条对角线满足 `i - j = d`，其中 `d = (m-1) - g`：`g=0` 时 `d=m-1`（只含左下角 `(m-1,0)`），随 `g` 增大 `d` 递减。

### 行范围
给定 `d`，需 `0<=i<=m-1` 且 `0<=j=i-d<=n-1`，解得 `i ∈ [max(0,d), min(m-1, n-1+d)]`，即 `lo..hi`。

### 交替方向
从左下角开始，第 `g` 条按 `g` 的奇偶决定从 `hi` 往 `lo`（偶）还是 `lo` 往 `hi`（奇），即可实现蛇形交替。取值 `mat[i][i-d]`。

### 复杂度
每个元素恰好访问一次，`O(m*n)` 时间、`O(1)` 额外空间（不计输出）。

```python
from typing import List
```

```python
# ✅ SOLUTION

class Solution:
    def diagonal_order(self, mat: List[List[int]]) -> List[int]:
        m = len(mat); n = len(mat[0]); res = []
        for g in range(m + n - 1):
            d = (m - 1) - g                       # cells on this diagonal have i - j = d
            lo = max(0, d); hi = min(m - 1, n - 1 + d)
            order = range(hi, lo - 1, -1) if g % 2 == 0 else range(lo, hi + 1)
            for i in order:
                res.append(mat[i][i - d])
        return res
```

```python
# Demo
sol = Solution()
print(sol.diagonal_order([[1,2,3],[4,5,6],[7,8,9]]))   # [7,4,8,9,5,1,2,6,3]
print(sol.diagonal_order([[1,2,3,4]]))                 # [1,2,3,4]
print(sol.diagonal_order([[1],[2],[3]]))               # [3,2,1]
```
