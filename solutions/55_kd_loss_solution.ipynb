```python
import torch
import torch.nn.functional as F
```

```python
# ✅ SOLUTION

def kd_loss(student_logits, teacher_logits, targets,
            temperature=2.0, alpha=0.5):
    teacher_probs = F.softmax(teacher_logits.detach() / temperature, dim=-1)
    student_log_probs = F.log_softmax(student_logits / temperature, dim=-1)
    kd = F.kl_div(student_log_probs, teacher_probs, reduction='batchmean') * (temperature ** 2)
    ce = F.cross_entropy(student_logits, targets)
    return alpha * kd + (1 - alpha) * ce
```

```python
student = torch.randn(4, 6)
teacher = torch.randn(4, 6)
targets = torch.randint(0, 6, (4,))
print('Loss:', kd_loss(student, teacher, targets))
```
