# Solution: Instance State Publish Operation

```python
# ✅ SOLUTION

class Solution:
    def find_operation(self, a: str, b: str) -> str:
        n = len(a)
        left = right = -1
        for i in range(n):
            if a[i] != b[i]:
                if left == -1:
                    left = i
                right = i
        if left == -1:            # A == B: no changing operation (guaranteed not to happen)
            return '-1'
        target = b[left]          # V is forced by the first differing cell
        if left > 0 and b[left - 1] == target:      # can extend left -> not unique
            return '-1'
        if right + 1 < n and b[right + 1] == target:  # can extend right -> not unique
            return '-1'
        return f'{left + 1} {right + 1} {target}'
```

```python
# Demo
sol = Solution()
for a, b in [('00000', '01110'), ('01000', '01110'), ('111111', '100001'), ('0', '1')]:
    print(a, '->', b, '=>', sol.find_operation(a, b))
```
