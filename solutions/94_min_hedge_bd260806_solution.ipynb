# 参考解法：最小对冲值

Reference solution.

## 解析

**结论：最平衡的拆分使异或最小，答案为 `(m // 2) XOR ((m + 1) // 2)`。**

### 从加法恒等式出发
对任意 `a, b`，`a + b = (a XOR b) + 2 * (a AND b)`。取 `a = y`、`b = m - y`，则 `a + b = m` 是常数，所以最小化 `a XOR b` 等价于最大化 `a AND b`。

### 最平衡拆分
`a AND b` 在 `a, b` 尽量接近时最大，即 `y = m // 2`、`m - y = (m + 1) // 2`。代入得答案 `(m//2) XOR ((m+1)//2)`。

### 特例
偶数 `m`：两半相等，异或为 0。`m = 2^k - 1`（全 1）：任何拆分都无进位，异或恒为 `m`，故最小值就是 `m`。

### 复杂度
`O(1)`；`m` 可达 `1e18`，Python 原生整数天然支持。

### 验证
已用 `min(y ^ (m-y))` 全枚举在数千组小 `m` 上对拍一致，并复现官方样例。

```python
# ✅ SOLUTION

class Solution:
    def min_hedge(self, m: int) -> int:
        return (m // 2) ^ ((m + 1) // 2)   # most balanced split minimizes the XOR
```

```python
sol = Solution()
print(sol.min_hedge(4))            # 0
print(sol.min_hedge(5))            # 1
print(sol.min_hedge(2 ** 60 - 1))  # 2**60 - 1  (all ones)
```
