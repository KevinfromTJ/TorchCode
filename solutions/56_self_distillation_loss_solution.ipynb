```python
import torch
import torch.nn.functional as F
```

```python
# ✅ SOLUTION

def self_distillation_loss(main_logits, aux_logits, targets,
                           temperature=2.0, ce_weight=1.0, distill_weight=0.5):
    ce = F.cross_entropy(main_logits, targets)
    teacher_probs = F.softmax(main_logits.detach() / temperature, dim=-1)
    aux_log_probs = F.log_softmax(aux_logits / temperature, dim=-1)
    kd = F.kl_div(aux_log_probs, teacher_probs, reduction='batchmean') * (temperature ** 2)
    return ce_weight * ce + distill_weight * kd
```

```python
main = torch.randn(4, 6)
aux = torch.randn(4, 6)
targets = torch.randint(0, 6, (4,))
print('Loss:', self_distillation_loss(main, aux, targets))
```
