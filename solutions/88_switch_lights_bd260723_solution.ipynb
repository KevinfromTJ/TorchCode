# 参考解法：开关控制灯

Reference solution.

## 解析

**结论：从左到右贪心。灯 `i` 是否需要按开关 `i`，完全由它当前状态（含左邻开关的进位）决定，累计按压次数即为答案。**

### 为什么是被迫的
处理到位置 `i` 时，开关 `1..i-1` 已经确定，能再改变灯 `i` 的只剩开关 `i`。因此若灯 `i` 此刻仍亮，就必须按开关 `i`；否则一定不能按（按了就没法在不影响已确定灯的前提下补救）。

### 进位传递
开关 `i` 会顺带切换灯 `i+1`，用 `prev` 记录这一进位。`cur = a[i] ^ prev` 表示灯 `i` 的真实当前状态。

### 复杂度
一次线性扫描，`O(n)` 时间、`O(1)` 额外空间。

```python
from typing import List
```

```python
# ✅ SOLUTION

class Solution:
    def min_ops(self, a: List[int]) -> int:
        ans = 0
        prev = 0                       # carry toggle from the previous switch
        for x in a:
            cur = x ^ prev
            if cur == 1:               # light still on -> must press this switch
                ans += 1
                prev = 1
            else:
                prev = 0
        return ans
```

```python
sol = Solution()
print(sol.min_ops([1, 1, 0, 1, 1]))   # 2
print(sol.min_ops([1, 1, 1]))         # 2
print(sol.min_ops([0, 0, 0]))         # 0
```
