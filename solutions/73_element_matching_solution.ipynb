# Solution: Multi-Constraint Element Matching

Reference solution — prefix counting with a hash map.


## 解析

**结论：固定右端点 `j`，用哈希表维护 `a` 的前缀出现次数。**

题目要求统计 `1 <= i <= j <= n` 且 `a[i] == b[c_j]` 的有序对 `(i, j)`。固定 `j` 后，令 `x = b[c_j]`，那么这个 `j` 的贡献就是**前缀 `a[1..j]` 中等于 `x` 的元素个数**。

从左到右扫描 `j`：

1. 先把 `a[j]` 计入哈希表（保证前缀范围是 `a[0..j]`，即 `i <= j`）；
2. 取 `x = b[c_j]`（`c_j` 是 1 下标，代码里用 `b[c[j]-1]`）；
3. `ans += cnt[x]`。

**易错点**：`c` 是 1 索引，务必减 1 转成 0 索引；先加入 `a[j]` 再查询，才能把 `i == j` 的情形算进去。

**复杂度**：每个位置只遍历一次，哈希表插入/查询均摊 `O(1)`，单组 `O(n)`；空间 `O(n)`。

```python
from typing import List
from collections import defaultdict
```

```python
# ✅ SOLUTION

class Solution:
    def count_pairs(self, a: List[int], b: List[int], c: List[int]) -> int:
        cnt = defaultdict(int)
        ans = 0
        for j in range(len(a)):
            cnt[a[j]] += 1              # prefix now covers a[0..j], so i <= j
            ans += cnt[b[c[j] - 1]]     # c[j] is 1-indexed -> b[c[j]-1]
        return ans
```

```python
# Demo
sol = Solution()
print(sol.count_pairs([1, 2, 1, 2, 1], [2, 1, 3, 1, 2], [2, 1, 5, 4, 3]))  # 5
print(sol.count_pairs([7, 7, 7], [1, 2, 7], [3, 3, 3]))                    # 6
```
