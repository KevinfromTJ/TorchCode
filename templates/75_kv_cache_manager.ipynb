# 🟡 Medium: Dynamic KV Cache Sparsification

> Algorithm problem (no PyTorch). LeetCode class-mode (stateful). Source: Huawei 2026-07-15.

In LLM autoregressive inference, the KV cache grows with sequence length. Implement a sparse KV cache with capacity `K`:

- `add(pos, score, key, value)`: insert a token (position, attention score, 4-dim Key, 4-dim Value). Positions are unique and strictly increasing. If the cache now exceeds `K`, **immediately evict the lowest-score token** (ties → evict the smaller position). Return the evicted position, or `None` if nothing was evicted.
- `query()`: return all cached tokens **sorted by position ascending**, as a list of `(pos, score, key, value)`.

### Signature
```python
from typing import List, Optional, Tuple

class Solution:
    def __init__(self, k: int): ...
    def add(self, pos: int, score: float, key: List[float], value: List[float]) -> Optional[int]: ...
    def query(self) -> List[Tuple[int, float, List[float], List[float]]]: ...
```

### Rules
- `1 <= K <= 1e5`, up to `1e5` operations; `0 <= score <= 1000`; Key/Value dim = 4.
- Target `O(log K)` per add (min-heap + lazy deletion), `O(K log K)` per query.
