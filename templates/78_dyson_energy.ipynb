# 🟡 Medium: Dyson Ring Energy Transport

> Algorithm problem (no PyTorch). LeetCode class-mode. Source: Pinduoduo 2026 Summer 4.12.

A ship starts at integer position `start` on a straight line with `fuel` units of fuel (moving 1 unit of distance costs 1 fuel; it may move left or right freely). There are `M` energy depots on the line; passing a depot's position **auto-collects** its energy. Under the fuel limit, maximise the total energy collected.

### Signature
```python
from typing import List

class Solution:
    def max_energy(self, start: int, fuel: int,
                   positions: List[int], energies: List[int]) -> int:
        ...
```

### Rules
- `positions[i]` is the coordinate of depot `i`, `energies[i]` its energy (energy > 0).
- Multiple depots may share a position; standing there collects all of them.
- The ship need not return anywhere; it just may not exceed `fuel` total distance.

*Hint:* an optimal route turns direction at most once, so collected depots form an interval `[L, R]` with `L <= start <= R`, costing `(R-L) + min(start-L, R-start)` fuel.
