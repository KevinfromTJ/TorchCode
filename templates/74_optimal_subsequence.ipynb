# 🔴 Hard: Optimal Subsequence (GCD Chain)

> Algorithm problem (no PyTorch). LeetCode class-mode. Source: Alibaba 2026-05-23.

Given an array `a` of length `n`, choose a length-`k` subsequence (indices `i_1 < i_2 < ... < i_k`) such that every adjacent pair satisfies `gcd(a[i_j], a[i_{j+1}]) > 1`. Among all valid subsequences, minimise the **maximum** element. Return that minimum possible maximum, or `-1` if no valid subsequence exists.

### Signature
```python

class Solution:
    def min_max_element(self, a: List[int], k: int) -> int:
        ...
```

### Rules
- `2 <= n <= 2e5`, `1 <= k <= n`, `1 <= a[i] <= 1e9`.
- Target `O(log n · Σ ω(a_i))` where `ω` is the number of distinct prime factors.

*Hint:* binary-search the answer; feasibility = "is there a length-`k` gcd-chain among elements `<= x`"; two numbers connect iff they share a prime factor.
