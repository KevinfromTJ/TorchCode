# 🟢 Easy: Nearest Free Dining Spot

> Algorithm problem (no PyTorch). LeetCode class-mode. Source: Alibaba 2026 Summer 4.18.

On the number line there are `n` forbidden **closed integer** intervals `[l_i, r_i]`. You are at integer position `p`. Find the nearest integer position `x` that is **not** inside any forbidden interval, and return `|x - p|`.

### Signature
```python
from typing import List

class Solution:
    def nearest_free_distance(self, p: int, intervals: List[List[int]]) -> int:
        ...
```

### Rules
- Intervals are closed: `x` is forbidden iff `l_i <= x <= r_i` for some `i`.
- Intervals may overlap; adjacent integer intervals (gap of 1) form one solid block.
- If `p` is already free, the answer is `0`.

*Hint:* merge overlapping/adjacent intervals; if `p` lands in block `[l, r]`, the nearest free integers are `l-1` and `r+1`.
