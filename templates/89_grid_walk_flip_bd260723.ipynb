# 🔴 Hard: 格子行走翻转

> 来源：百度秋招 2026-07-23（第 3 题）

有一条从 1 开始编号的无限一维格子带。初始时**素数编号**的格子为黑色（记 `1`），其余为白色（记 `0`）。共有 `n` 个人；第 `i` 个人（从 1 开始）沿二进制字符串 `s` 的前 `i` 个字符行走。每次行走都从格子 1 出发，遇到字符 `'0'` 就跳到当前格子**右侧最近的白色格子**，遇到 `'1'` 就跳到右侧最近的黑色格子。走完后，这个人会把落脚的格子**翻转颜色**。颜色改变会保留并影响后续的人。

请返回最终颜色与初始颜色**不同**的格子：其数量，以及这些格子编号的升序列表 `(count, sorted_list)`。

```python
        ...
```

*提示：* 只需在素数底色之上维护每个格子的**翻转奇偶**：`color(g) = is_prime(g) XOR parity[g]`。逐个模拟 `n` 个人：从 `cur=1` 出发，对每个字符向右扫描（`g = cur+1, cur+2, ...`）找到第一个颜色匹配目标（`'0'` 找白、`'1'` 找黑）的格子并移动过去；走完后翻转落脚格的奇偶。最后输出奇偶为 1 的格子（升序）。
