> Source: Pinduoduo 2026-07-02 (Problem 1).

Traverse an `m x n` matrix by **anti-diagonals starting from the bottom-left** corner, alternating the scan direction on successive diagonals. Return the visited values in order.

### Signature
```python

class Solution:
    def diagonal_order(self, mat: List[List[int]]) -> List[int]:
        ...
```

Example `[[1,2,3],[4,5,6],[7,8,9]]` -> `[7,4,8,9,5,1,2,6,3]`.

*Hint:* there are `m+n-1` diagonals. On diagonal `g` the cells satisfy `i - j = d` with `d = (m-1) - g`; walk `i` from `hi` down to `lo` on even `g` and up on odd `g` (`lo=max(0,d)`, `hi=min(m-1, n-1+d)`).
