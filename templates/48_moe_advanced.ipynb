### Requirements
- Sparse routing: only top-k experts per token
- Top-k weights use `softmax(topk_logits)`
- Aux loss (Switch style):
  `importance = softmax(router_logits).mean(...)`
  `load = top1 expert frequency`
  `aux = num_experts * sum(importance * load)`
