# 🔴 Hard: GDPO Loss

Implement a full **GDPO** loss with **multi-reward decoupled normalization**.

This version assumes `rewards` has shape `(B, R)` where each column is a different reward source.

### Signature
```python
from torch import Tensor

def gdpo_loss(logps: Tensor, old_logps: Tensor, ref_logps: Tensor,
              rewards: Tensor, group_ids: Tensor, completion_mask: Tensor,
              reward_weights: Tensor | None = None,
              clip_ratio: float = 0.2, beta: float = 0.1, eps: float = 1e-5) -> Tensor:
    # logps / old_logps / ref_logps: (B, T)
    # rewards: (B, R)
```

### Rules
- Normalize each reward dimension independently within each group
- Aggregate reward dimensions with `reward_weights` (or equal weights by default)
- Batch-normalize the aggregated advantage one more time
- Use masked GRPO-style clipped loss + reference KL
