# 🟢 Easy: 开关控制灯

> 来源：百度秋招 2026-07-23（第 1 题）

有 `n` 盏灯和 `n` 个开关，灯与开关均从 1 开始编号，初始状态由数组 `a` 给出（`1` 亮、`0` 灭）。按下开关 `i`（`1 ≤ i < n`）会同时切换灯 `i` 和灯 `i+1` 的状态；按下开关 `n` 只切换灯 `n`。求把所有灯都关掉所需的**最少按压次数**。

### 函数签名
```python
from typing import List

class Solution:
    def min_ops(self, a: List[int]) -> int:
        ...
```

*提示：* 从左往右扫描。灯 `i` 之后只能由开关 `i` 再影响（开关 `i-1` 已定），所以它的决定是被迫的：维护上一个开关带来的进位 `prev`，`cur = a[i] XOR prev`，若 `cur==1` 必须按开关 `i`（答案 +1，进位置 1），否则进位置 0。贪心即最优，`O(n)`。
