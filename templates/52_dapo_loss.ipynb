```python
              clip_ratio: float = 0.2, clip_ratio_high: float = 0.28,
              beta: float = 0.1, eps: float = 1e-5) -> Tensor:
    # logps / old_logps / ref_logps: (B, T)
    # rewards: (B,)
```

### Rules
- Normalize rewards within each group
- Use token-level ratios like GRPO
- Use asymmetric clipping: lower `1 - clip_ratio`, upper `1 + clip_ratio_high`
- Normalize over all active tokens in the batch, not per sequence
