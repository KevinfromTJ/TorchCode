### Requirements
- Token embedding + **learnable** positional embedding.
- `n_layer` pre-norm blocks: `x = x + attn(ln1(x))`, `x = x + mlp(ln2(x))`.
- Self-attention must be **causal** (a token only attends to itself and earlier positions).
- MLP: `Linear(n_embd, 4*n_embd) -> GELU -> Linear(4*n_embd, n_embd)`.
- Final `LayerNorm`, then `lm_head = Linear(n_embd, vocab_size)`.
- Loss = `F.cross_entropy` over flattened logits/targets.
- `generate` crops context to the last `block_size` tokens each step.
