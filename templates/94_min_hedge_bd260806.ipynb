# 🟡 Medium: 最小对冲值

> 来源：百度机考算法岗 2026-08-06（第 2 题）

把一个非负整数 `m` 拆成 `y` 与 `m - y`（`0 ≤ y ≤ m`）。「对冲值」定义为 `y XOR (m - y)`。返回可以取到的最小对冲值。

### 函数签名
```python
class Solution:
    def min_hedge(self, m: int) -> int:
        ...
```

*提示：* 由 `a + b = (a XOR b) + 2*(a AND b)`（取 `a = y, b = m - y`），最小化 `a XOR b` 等价于最大化 `a AND b`，而这在**最平衡**的拆分处取得。因此答案为 `(m // 2) XOR ((m + 1) // 2)`：偶数 `m` 得 0，`m = 2^k - 1` 得 `m` 本身。每次查询 `O(1)`，`m` 可达 `1e18`（用 64 位）。
