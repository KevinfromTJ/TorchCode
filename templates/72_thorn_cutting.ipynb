# 🟢 Easy: Thorn Forest Optimal Cutting

> Algorithm problem (no PyTorch). LeetCode class-mode. Source: Alibaba 2026-05-23.

There are `n` thorns; the `i`-th has hardness `a[i]`. A blade starts with sharpness `K`. You may attempt any subset of thorns in any order (each at most once):

- if the current sharpness `K >= a[i]`, the thorn is cut successfully;
- **regardless of success, every attempt permanently reduces sharpness by 1**;
- you may skip any thorn for free.

Return the maximum number of thorns you can successfully cut.

### Signature
```python
from typing import List

class Solution:
    def max_cuts(self, a: List[int], k: int) -> int:
        ...
```

### Rules
- `1 <= n <= 2e5`, `1 <= K <= 1e9`, `1 <= a[i] <= 1e9`.
- Target `O(n log n)`.
