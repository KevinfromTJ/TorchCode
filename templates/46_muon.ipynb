# 🔴 Hard: Muon Optimizer (Simplified)

Implement a simplified **Muon** optimizer (momentum + Newton-Schulz orthogonalized updates).

### Signature
```python
class MyMuon:
    def __init__(self, params, lr=2e-2, beta=0.95, weight_decay=0.0, ns_steps=5, eps=1e-8): ...
    def step(self): ...
    def zero_grad(self): ...
```

### Rules
- Momentum: `m = beta*m + (1-beta)*(grad + weight_decay * param)`
- For tensors with `ndim >= 2`: orthogonalize `m` using Newton-Schulz
- For vectors (`ndim == 1`): use raw momentum directly
- Update: `param -= lr * update`
