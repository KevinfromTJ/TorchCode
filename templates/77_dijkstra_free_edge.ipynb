# 🟡 Medium: Shortest Path with One Free Edge

> Algorithm problem (no PyTorch). LeetCode class-mode.

You are given a **directed weighted** graph with `N` nodes (numbered `1..N`) and a list of edges `(u, v, w)` meaning a directed edge `u -> v` with weight `w`. **All weights are positive.**

You may pick **at most one** edge and set its weight to `0` (a single free edge). Find the minimum total weight of a path from node `1` to node `N`. If no path exists, return `-1`.

### Signature
```python

    def min_weight_path(self, n: int, edges: List[Tuple[int, int, int]]) -> int:
        ...
```

### Rules
- Weights are strictly positive, so Dijkstra applies.
- The free edge is optional (using it never hurts, but you may use zero edges).
- If `N == 1`, the answer is `0` (empty path).
- There may be parallel edges and cycles.

*Hint:* run Dijkstra over doubled states `(node, used)` where `used ∈ {0,1}` marks whether the free edge has already been spent.
