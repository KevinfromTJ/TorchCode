# 🟢 Easy: 括号分层消除

> 来源：百度机考算法岗 2026-07-30（第 1 题）

给定一个合法（左右匹配）的括号字符串。每一轮同时删除**所有最内层**的括号对 `()`，直到字符串为空。对原串中的每个字符，求它在第几轮被删除，按原始顺序返回列表。

### 函数签名
```python
from typing import List

class Solution:
    def rounds(self, s: str) -> List[int]:
        ...
```

*提示：* 消除轮次只取决于嵌套深度。一次扫描：遇 `'('` 先把深度 +1 再记录，遇 `')'` 先记录当前深度再 -1。设 `D` 为最大记录深度，深度为 `d` 的字符在第 `D - d + 1` 轮消失（最深层 `d = D` 最先消失）。`O(n)`。
