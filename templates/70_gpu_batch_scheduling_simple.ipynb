# 🟡 Medium: GPU Batch Scheduling (No Discards)

> Algorithm problem (no PyTorch). LeetCode class-mode. Simplified variant of problem 67 — **no request may be dropped**.

An inference engine batches requests to improve GPU utilization. A batch must satisfy:

1. at most `C` requests per batch;
2. within a batch, `max_len - min_len <= K` (padding waste limit);
3. requests may be reordered freely.

**Every** request must be served (unlike problem 67, there is no drop budget). Given the token lengths, return the **minimum number of batches**.

### Signature
```python

        ...
```

### Rules
- `1 <= N <= 1e5`, `0 <= K <= 1e9`, `1 <= C <= N`, `1 <= L_i <= 1e9`.
- Every request is served, so the answer is between `ceil(N/C)` and `N`.
- Target `O(N log N)` (sort + single greedy sweep).
