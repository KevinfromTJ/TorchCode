# 🟢 Easy: Binary Cross-Entropy Loss

Implement numerically stable **binary cross-entropy loss with logits**.

$$
\mathcal{L}(x, y) = -y \log \sigma(x) - (1-y) \log (1-\sigma(x))
$$

### Signature
```python
def bce_loss(logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
    # logits, targets: same shape
    # targets can be 0/1 or soft labels in [0, 1]
    # Returns: scalar mean loss
```

### Rules
- Treat inputs as **logits**, not probabilities
- Do **NOT** use `F.binary_cross_entropy_with_logits`
- Must stay stable for very large positive/negative logits

````python
import torch
````

````python
# ✏️ YOUR IMPLEMENTATION HERE

def sigmoid(x):
    # x = torch.clip(x, min=200,max=200)
    return 1 / (1+(-x).exp())
def bce_loss(logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
    # pass
    return (- targets * sigmoid(logits).log() - (1-targets) * (1-sigmoid(logits)).log()).mean()
    

````

````python
# 🧪 Debug
logits = torch.tensor([2.0, -1.0, 0.5, -3.0])
targets = torch.tensor([1.0, 0.0, 1.0, 0.0])
print('Loss:', bce_loss(logits, targets))
print('Ref: ', torch.nn.functional.binary_cross_entropy_with_logits(logits, targets))
````
