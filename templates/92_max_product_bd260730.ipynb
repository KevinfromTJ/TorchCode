给定一个正整数序列。你必须**恰好执行 `k` 次操作**，每次操作把某个当前 `> 1` 的元素减 1（若所有元素都等于 1 则提前停止）。要使序列的乘积最大，返回该最大乘积对 `1e9+7` 取模的结果。

### 函数签名

```python
        ...
```

*提示：* 贪心：每次都减当前**最大**的元素。因为当 `x ≥ y` 时 `(x-1)/x ≥ (y-1)/y`，减最大元素对乘积的相对损失最小。升序排序，`total = sum(x-1)`；若 `k ≥ total` 答案为 1。否则从高处把最大的 `cnt` 个元素一层层压低（把 `cnt` 个降到下一档的代价为 `(level - next) * cnt`）；对剩余 `k` 做 `q, r = divmod(k, cnt)`，`high = level - q`，乘积 = 未动前缀 × `high^(cnt-r)` × `(high-1)^r`，全程取模。
