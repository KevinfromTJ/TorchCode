# 🟡 Medium: Magic Vine Cutting

> Algorithm problem (no PyTorch). LeetCode class-mode. Source: Alibaba 2026 Summer 7.23.

A vine is `n` magic segments `a[1..n]`. The **remaining vine is always a prefix**. Each operation is one of:

1. Find the **last occurrence** of the current prefix's **maximum** hardness; cut that position and everything to its right.
2. Find the **last occurrence** of the current prefix's **minimum** hardness; cut that position and everything to its right.

"Last occurrence" = on ties, take the largest index. Cutting at position `pos` leaves a prefix of length `pos-1`. Return the **minimum number of operations** needed until segment 1 is also cut (prefix length becomes 0).

*Hint:* `dp[k]` = min ops to fully cut a prefix of length `k`. For length `k`, one op jumps to `maxpos[k]-1` or `minpos[k]-1`, so `dp[k] = 1 + min(dp[maxpos[k]-1], dp[minpos[k]-1])`.
