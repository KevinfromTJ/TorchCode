# 🟡 Medium: Side-Scroller Coin Catching

> Algorithm problem (no PyTorch). LeetCode class-mode.

$$T_B - T_A > |X_B - X_A|$$

(strictly greater — a short pickup stun). You may start anywhere. What is the **maximum number of coins** you can catch?

> If several coins drop at the same time **and** coordinate, you can take at most one.
