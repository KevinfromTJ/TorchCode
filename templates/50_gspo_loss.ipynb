# 🔴 Hard: GSPO Loss

Implement a full **GSPO (Group Sequence Policy Optimization)** loss.

Compared with GRPO, GSPO uses a **sequence-level importance ratio** instead of a token-level ratio.

### Signature
```python
from torch import Tensor

def gspo_loss(logps: Tensor, old_logps: Tensor, ref_logps: Tensor,
              rewards: Tensor, group_ids: Tensor, completion_mask: Tensor,
              clip_ratio: float = 0.2, beta: float = 0.1, eps: float = 1e-5) -> Tensor:
    # logps / old_logps / ref_logps: (B, T)
    # rewards: (B,)
    # group_ids: (B,)
    # completion_mask: (B, T)
```

### Rules
- Normalize rewards within each group
- Build one ratio per sequence using only valid tokens in `completion_mask`
- Apply clipping once per sequence
- Mask KL / valid-token averaging correctly
- Detach old policy, reference policy, and reward-derived advantages
