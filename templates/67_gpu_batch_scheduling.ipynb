# 🔴 Hard: GPU Batch Scheduling (Min Batches)

> Algorithm problem (no PyTorch). LeetCode class-mode.

An inference engine batches requests to improve GPU utilization. A batch must satisfy:

1. at most `C` requests per batch;
2. within a batch, `max_len - min_len <= K` (padding waste limit);
3. requests may be reordered freely.

To avoid outliers dragging down throughput, you may **drop at most `M`** requests entirely. Given the token lengths, return the **minimum number of batches** needed to serve the kept requests.

### Signature
```python
from typing import List

class Solution:
    def min_batches(self, lengths: List[int], m: int, k: int, c: int) -> int:
        ...
```

### Rules
- `1 <= N <= 1e5`, `0 <= M <= 50`, `0 <= K <= 1e9`, `1 <= C <= N`, `1 <= L_i <= 1e9`.
- Dropping all requests is allowed (answer can be `0`).
- Target `O(N log N + N*M)`.
