### 函数签名

```python
        ...
```

*提示：* 在**倍长数组**上用前缀余数：`P[0]=0`，`P[i]=(P[i-1]+a[(i-1)%n]) % M`。起点 `s` 的轨迹点恰好等于 `P[s+1..s+t]`，其两两不同等价于这段 `P` 值两两不同。于是 `L_s` = 从下标 `s+1` 开始 `P` 中不重复值的最长连续段长度（上限 `n`）。用带 `seen` 集合与单调右指针的滑动窗口对 `j = s+1`（`j` 取 1..n）累加 `(r - j)`，`O(n)`。
