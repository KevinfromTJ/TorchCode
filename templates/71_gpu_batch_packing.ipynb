# 🟢 Easy: GPU Batch Packing (Fixed Token Buffer)

> Algorithm problem (no PyTorch). LeetCode class-mode. Variant of 67 / 70 — continuous-batching style token packing.

Requests **arrive in the given order** and are packed into fixed-size batches (no dropping, no reordering):

1. each batch is a buffer of exactly `B` tokens; unused space is padded with empty tokens;
2. a request of length `L` consumes `L` tokens;
3. inside a batch, **every two consecutive requests are separated by `G` pad tokens**.

So a batch holding `m` requests occupies `sum(L) + G*(m - 1) <= B` tokens. Pack each incoming request into the current batch if it still fits; otherwise start a new batch. Return the **number of batches** produced.

### Rules
- `1 <= N <= 1e5`, `1 <= B <= 1e9`, `0 <= G <= 1e9`, `1 <= L_i <= B` (every request fits an empty buffer).
- Requests are served in the given order (no reordering) and none are dropped.
- Target `O(N)`.
