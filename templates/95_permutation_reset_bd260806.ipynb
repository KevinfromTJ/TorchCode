```python
        ...
```

*提示：* 把 `p` 分解为若干不相交的**循环**。每个循环上的字符构成一个环形串，每次操作旋转一步，因此它经过其最小旋转周期 `d` 后复位（`d` 是循环长度 `L` 的最小约数，使得该环形串恰好由长度 `d` 的块重复 `L/d` 次；全部字符相同则 `d = 1`）。整个串复位于所有循环周期的 **LCM**。用 `gcd` 累积 LCM 并对 `1e9+7` 取模（真实 LCM 可能超过 64 位）。
