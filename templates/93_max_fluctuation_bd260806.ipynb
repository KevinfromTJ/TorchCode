# 🟢 Easy: 波动贡献最大化

> 来源：百度机考算法岗 2026-08-06（第 1 题）

把整个序列划分为若干**非空连续**段，且完整覆盖序列。一个段的「波动贡献」为 `(max - min) * 段长度`。返回所有划分方式中，总贡献的最大值。

### 函数签名
```python
from typing import List

class Solution:
    def max_total(self, v: List[int]) -> int:
        ...
```

*提示：* 对任意段都有 `(max - min) ≤ (全局max - 全局min)`，故每段贡献 `≤ (M - m) * 段长`。对某个划分求和得上界 `(M - m) * n`，而**不切分（整段作为一段）**恰好取到这个上界。所以答案就是 `(max(v) - min(v)) * n`。注意用 64 位。
